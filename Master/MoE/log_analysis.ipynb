```python
import re
from collections import defaultdict

# Caminho do arquivo
filepath = "log_training.txt"

# Regex para capturar: "Expert: NOME | Weight: VALOR"
pattern = r"Expert:\s*(.*?)\s*\|\s*Weight:\s*([0-9.]+)"

expert_counts = defaultdict(int)
expert_weights = defaultdict(list)

with open(filepath, "r", encoding="utf-8") as f:
    for line in f:
        match = re.search(pattern, line)
        if match:
            expert_name = match.group(1)
            weight = float(match.group(2))
            
            # Contabiliza
            expert_counts[expert_name] += 1
            expert_weights[expert_name].append(weight)

# Total de escolhas
total_selections = sum(expert_counts.values())

print("=== Estatísticas dos Experts ===\n")

for expert, count in expert_counts.items():
    pct = (count / total_selections) * 100
    avg_weight = sum(expert_weights[expert]) / len(expert_weights[expert])
    
    print(f"Expert: {expert}")
    print(f"  - Seleções: {count} ({pct:.2f}%)")
    print(f"  - Peso médio: {avg_weight:.4f}\n")
```

```python
import re
import pandas as pd
from collections import defaultdict

# Caminho do arquivo
filepath = "./2_many_models_std/log_training_new.txt"

# Regex
start_exp_pattern = re.compile(r"STARTING TRAINING")
model_path_pattern = re.compile(r"trained_models/horizon_(\d+)/([^/]+)/")
epoch_pattern = re.compile(
    r"Epoch\s+(\d+)/(\d+)\s+\|\s+Train Loss:\s+([0-9]*\.?[0-9]+)"
)

experiments = []
current_experiment = None

with open(filepath, "r", encoding="utf-8") as f:
    for line in f:
        # Novo experimento
        if start_exp_pattern.search(line):
            if current_experiment is not None:
                experiments.append(current_experiment)

            current_experiment = {
                "epoch_loss": {},   # epoch -> loss
                "horizon": None,
                "state": None
            }
            continue

        # Caminho do modelo (horizonte + estado)
        model_match = model_path_pattern.search(line)
        if model_match and current_experiment is not None:
            current_experiment["horizon"] = int(model_match.group(1))
            current_experiment["state"] = model_match.group(2)
            continue

        # Informações das épocas
        epoch_match = epoch_pattern.search(line)
        if epoch_match and current_experiment is not None:
            epoch = int(epoch_match.group(1))
            loss = float(epoch_match.group(3))
            current_experiment["epoch_loss"][epoch] = loss

# Último experimento
if current_experiment is not None:
    experiments.append(current_experiment)

# =========================
# Estatísticas + DataFrame
# =========================

rows = []

print(f"Total de experimentos encontrados: {len(experiments)}\n")

for i, exp in enumerate(experiments, 1):
    epoch_loss = exp["epoch_loss"]

    if not epoch_loss:
        print(f"Experimento {i}: nenhum dado de época encontrado\n")
        continue

    epochs = list(epoch_loss.keys())
    losses = list(epoch_loss.values())

    n_epochs = len(epochs)
    min_epoch = min(epochs)
    max_epoch = max(epochs)

    avg_loss = sum(losses) / n_epochs
    final_loss = epoch_loss[max_epoch]

    print(f"Experimento {i}")
    print(f"  Horizonte: {exp['horizon']}")
    print(f"  Estado: {exp['state']}")
    print(f"  Épocas registradas: {n_epochs}")
    print(f"  Intervalo de épocas: {min_epoch} → {max_epoch}")
    print(f"  Média Train Loss: {avg_loss:.4f}")
    print(f"  Última época: {max_epoch}")
    print(f"  Final Train Loss: {final_loss:.4f}")
    print("-" * 60)

    # Linha do DataFrame (APENAS o que você pediu)
    rows.append({
        "horizon": exp["horizon"],
        "state": exp["state"],
        "num_epochs": n_epochs,
        "final_epoch": max_epoch,
        "final_train_loss": final_loss,
        "mean_train_loss": avg_loss
    })

# Criação do DataFrame
df = pd.DataFrame(rows)

print("\nDataFrame final:")
print(df.head())
```

```python
import pandas as pd

# ================================
# Supondo que seu DataFrame seja df
# ================================

# Total de casos
total_cases = len(df)

print("=" * 80)
print(f"TOTAL DE CASOS NO DATAFRAME: {total_cases}")
print("=" * 80, "\n")

# ================================
# 1. Criar faixas de final_epoch
# ================================
bins = [0, 5, 10, df['final_epoch'].max()]
labels = ['1-5', '6-10', '>10']

df['epoch_range'] = pd.cut(
    df['final_epoch'],
    bins=bins,
    labels=labels,
    include_lowest=True
)

print("Faixas de epochs criadas:")
print(labels)
print("\n")

# ================================
# 2. Contagem e porcentagem por faixa
# ================================
epoch_summary = (
    df.groupby('epoch_range')
      .size()
      .reset_index(name='num_cases')
)

epoch_summary['percentage'] = (
    epoch_summary['num_cases'] / total_cases * 100
)

print("=" * 80)
print("DISTRIBUIÇÃO DE CASOS POR FAIXA DE FINAL_EPOCH")
print("=" * 80)
print(epoch_summary.to_string(index=False))
print("\n")

# ================================
# 3. Função auxiliar para imprimir detalhes
# ================================
def print_epoch_range_details(df, epoch_label):
    subset = df[df['epoch_range'] == epoch_label]

    print("-" * 80)
    print(f"DETALHES — FINAL_EPOCH ENTRE {epoch_label}")
    print(f"Quantidade de casos: {len(subset)} "
          f"({len(subset) / total_cases * 100:.2f}% do total)")
    print("-" * 80)

    if subset.empty:
        print("Nenhum caso encontrado.\n")
        return

    print(
        subset[['horizon', 'final_epoch', 'final_train_loss']]
        .sort_values('final_epoch')
        .to_string(index=False)
    )
    print("\n")

# ================================
# 4. Imprimir detalhes por faixa
# ================================
print_epoch_range_details(df, '1-5')
print_epoch_range_details(df, '6-10')
print_epoch_range_details(df, '>10')

print("=" * 80)
print("FIM DA ANÁLISE")
print("=" * 80)
```

```python
rows = []

for horizon, df_h in df.groupby("horizon"):
    total_exps = len(df_h)

    summary = (
        df_h
        .groupby("final_epoch")
        .agg(
            quantidade=("final_epoch", "count"),
            mean_final_train_loss=("final_train_loss", "mean")
        )
        .reset_index()
    )

    for _, row in summary.iterrows():
        quantidade = int(row["quantidade"])
        percentual_total = (quantidade / total_exps) * 100

        rows.append({
            "Horizonte": horizon,
            "Época": int(row["final_epoch"]),
            "Quantidade": quantidade,
            "Percentual total": round(percentual_total, 4),
            "Média Loss": row["mean_final_train_loss"]
        })

# DataFrame final
df_summary = pd.DataFrame(rows)

# Ordenação (opcional, mas recomendada)
df_summary = df_summary.sort_values(
    ["Horizonte", "Época"]
).reset_index(drop=True)

# Exporta para Excel
output_path = "summary_final_epoch_by_horizon.xlsx"
df_summary.to_excel(output_path, index=False)

print(f"Excel gerado com sucesso: {output_path}")
df_summary
```
